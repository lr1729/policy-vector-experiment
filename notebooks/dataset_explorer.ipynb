```python
import json
from pathlib import Path
from itertools import islice

DATA_PATH = Path('data/on_policy_persona.json')
with open(DATA_PATH) as f:
    dataset = json.load(f)
examples = dataset['examples']
len(examples)
```

```python
def show_example(example, pair_limit=1):
    snippet = example['prompt'][:200]
    if len(example['prompt']) > 200:
        snippet += '...'
    print(f"Example ID: {example['example_id']}")
    print(f"Prompt: {snippet}")
    print(f"On-policy variants: {len(example['on_policy'])}; Off-policy variants: {len(example['off_policy'])}")
    print()
    for idx, (on, off) in enumerate(islice(zip(example['on_policy'], example['off_policy']), pair_limit), 1):
        sim = off['metadata'].get('similarity', 'N/A')
        on_lines = on['reasoning'].splitlines()
        off_lines = off['reasoning'].splitlines()
        on_first = on_lines[1] if len(on_lines) > 1 else ''
        off_first = off_lines[1] if len(off_lines) > 1 else ''
        answers_match = on['final_answer'].strip() == off['final_answer'].strip()
        print(f"  Pair {idx}")
        print(f"    Similarity: {sim}")
        print(f"    On reasoning (first line): {on_first}")
        print(f"    Off reasoning (first line): {off_first}")
        print(f"    Final answer matches: {answers_match}")
        print()
    print('-' * 80)
```

```python
show_example(examples[0], pair_limit=3)
```

```python
for ex in examples[1:4]:
    show_example(ex, pair_limit=1)
```
